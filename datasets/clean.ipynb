```python
import pandas as pd
# Пути к файлам
FILE_PATH = "../../../datasets/electricity_75.csv"
OUTPUT_PATH = "../../../datasets/electricity_75_cleaned.csv"

# Загружаем исходный CSV
df = pd.read_csv(FILE_PATH)

print("Исходные данные:")
display(df.head())

# Столбцы с потреблением и генерацией (МВт*ч)
consumption_columns = [
    "planned_consumption",
    "actual_consumption",
    "planned_generation",
    "actual_generation"
]

# Очищаем столбцы потребления и генерации
for col in consumption_columns:
    df[col] = (
        df[col]
        .astype("string")
        .str.replace(r"МВт\s*\\?\*\s*ч", "", regex=True)
        .str.replace(r"\./\s*МВт\s*\\?\*\s*ч", "", regex=True)
        .str.replace(r"\s+", "", regex=True)
        .str.strip()
    )

    df[col] = pd.to_numeric(df[col], errors="coerce")

# Очищаем столбец цены (руб./МВт*ч)
df["average_price"] = (
    df["average_price"]
    .astype("string")
    .str.replace(r"руб\./МВт\s*\\?\*\s*ч", "", regex=True)
    .str.replace(r"\s+", "", regex=True)
    .str.strip()
)

df["average_price"] = pd.to_numeric(
    df["average_price"],
    errors="coerce"
)

# Сохраняем очищенные данные в новый CSV
df.to_csv(
    OUTPUT_PATH,
    index=False,
    encoding="utf-8-sig"
)

print("\nОчищенные данные:")
display(df.head())

print(f"\nФайл сохранён: {OUTPUT_PATH}")

# Проверяем типы данных
print("\nТипы данных после очистки:")
print(df[consumption_columns + ["average_price"]].dtypes)
```
